```python
import numpy as np
import matplotlib.pyplot as plt

E_t0 = np.loadtxt('E_step0.txt')
E_tmid = np.loadtxt('E_step5000.txt')
E_tend = np.loadtxt('E_step10000.txt')

NX = len(E_t0)
DX = 1.0
x = np.arange(NX) * DX

plt.figure(figsize=(10, 6))

plt.plot(x, E_t0, label='Time step 0')
plt.plot(x, E_tmid, label='Time step NSTEPS/2')
plt.plot(x, E_tend, label='Time step NSTEPS')

plt.xlabel('Position')
plt.ylabel('Electric Field E')
plt.title('Electric Field at Different Time Steps')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
data = np.loadtxt("E_step10000.txt")
peak_sim = np.max(np.abs(data))
peak_ref = 0.5
error = np.abs(peak_sim - peak_ref) / peak_ref * 100
print(f"Relative Error: {error:.2f}%")
```
